```python
import re
import random

random.seed(42)

# Load all three datasets
with open("/content/1661-0.txt", "r", encoding="utf-8-sig") as f:
    sherlock_text = f.read()

with open("/content/alice.txt", "r", encoding="utf-8-sig") as f:
    alice_text = f.read()

with open("/content/pride.txt", "r", encoding="utf-8-sig") as f:
    pride_text = f.read()


def extract_book_text(raw_text):
    start_pattern = r"\*\*\*\s*START OF THE PROJECT GUTENBERG EBOOK.*?\*\*\*"
    end_pattern = r"\*\*\*\s*END OF THE PROJECT GUTENBERG EBOOK.*?\*\*\*"

    start = re.search(start_pattern, raw_text, re.IGNORECASE)
    end = re.search(end_pattern, raw_text, re.IGNORECASE)

    if start and end:
        return raw_text[start.end():end.start()]

    return raw_text


def clean_and_split(raw_text):
    text = extract_book_text(raw_text)

    # Remove illustration placeholders
    text = re.sub(r"\[Illustration[^\]]*\]", " ", text, flags=re.IGNORECASE)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # Split into sentences
    parts = re.split(r'(?<=[.!?])\s+', text)

    cleaned = []

    for sentence in parts:
        sentence = sentence.strip()

        if len(sentence.split()) < 5:
            continue

        if re.match(
            r"^(CHAPTER|CONTENTS|THE ADVENTURES OF SHERLOCK HOLMES)",
            sentence,
            re.IGNORECASE
        ):
            continue

        cleaned.append(sentence)

    return cleaned


# Process each book
sherlock_sentences = clean_and_split(sherlock_text)
alice_sentences = clean_and_split(alice_text)
pride_sentences = clean_and_split(pride_text)

print("Sherlock Holmes:", len(sherlock_sentences))
print("Alice in Wonderland:", len(alice_sentences))
print("Pride and Prejudice:", len(pride_sentences))

# Combine all sentences
all_sentences = (
    sherlock_sentences
    + alice_sentences
    + pride_sentences
)

# Shuffle reproducibly
random.shuffle(all_sentences)

# Select up to 10,000 sentences
sentences = all_sentences[:10000]

print("\nCombined dataset ready!")
print("Total sentences:", len(sentences))

print("\nFirst 10 sentences:\n")

for i, sentence in enumerate(sentences[:10], 1):
    print(f"{i}. {sentence}")
```

```python
import re
from collections import Counter

# Normalize text
processed_sentences = []

for sentence in sentences:
    sentence = sentence.lower().strip()

    # Keep words and basic punctuation
    tokens = re.findall(r"\b[a-z]+(?:'[a-z]+)?\b", sentence)

    if len(tokens) >= 3:
        processed_sentences.append(tokens)

# Build vocabulary
word_counts = Counter(
    word
    for sentence in processed_sentences
    for word in sentence
)

# Reserve 0 for padding and 1 for unknown words
word_to_id = {
    "<PAD>": 0,
    "<UNK>": 1
}

for word, count in word_counts.most_common():
    word_to_id[word] = len(word_to_id)

# Reverse mapping
id_to_word = {idx: word for word, idx in word_to_id.items()}

print("Tokenization completed!")
print("Total sentences:", len(processed_sentences))
print("Total word occurrences:", sum(word_counts.values()))
print("Vocabulary size:", len(word_to_id))

print("\nSample tokenized sentence:")
print(processed_sentences[0])

print("\nSample word-to-ID mapping:")
for word, idx in list(word_to_id.items())[:20]:
    print(f"{word}: {idx}")
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

# Generate input-target sequences
sequences = []

for tokens in processed_sentences:
    token_ids = [word_to_id[word] for word in tokens]

    # Create progressively longer sequences
    for i in range(1, len(token_ids)):
        sequences.append(token_ids[:i + 1])

print("Total training sequences:", len(sequences))

# Find the maximum sequence length
max_sequence_length = max(len(seq) for seq in sequences)

print("Maximum sequence length:", max_sequence_length)

# Pad sequences to equal length
padded_sequences = pad_sequences(
    sequences,
    maxlen=max_sequence_length,
    padding="pre",
    truncating="pre"
)

# Separate inputs and targets
X = padded_sequences[:, :-1]
y = padded_sequences[:, -1]

print("\nInput shape:", X.shape)
print("Target shape:", y.shape)

# Display one example
sample_input = X[0]
sample_target = y[0]

print("\nSample input IDs:", sample_input)
print("Target ID:", sample_target)

print("Input words:",
      [id_to_word[i] for i in sample_input if i != 0])

print("Target word:", id_to_word[sample_target])
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.sequence import pad_sequences

MAX_LEN = 20

# Split at sentence level first
train_sentences, test_sentences = train_test_split(
    processed_sentences,
    test_size=0.2,
    random_state=42
)

def create_sequences(sentence_list):
    sequences = []

    for tokens in sentence_list:
        token_ids = [word_to_id.get(word, word_to_id["<UNK>"]) for word in tokens]

        # Generate next-word prediction sequences
        for i in range(1, len(token_ids)):
            sequence = token_ids[max(0, i - MAX_LEN + 1):i + 1]
            sequences.append(sequence)

    # Pad sequences to a fixed length
    padded = pad_sequences(
        sequences,
        maxlen=MAX_LEN,
        padding="pre",
        truncating="pre"
    )

    X = padded[:, :-1]
    y = padded[:, -1]

    return X, y

X_train, y_train = create_sequences(train_sentences)
X_test, y_test = create_sequences(test_sentences)

print("Training sentences:", len(train_sentences))
print("Testing sentences:", len(test_sentences))
print("Training input shape:", X_train.shape)
print("Training target shape:", y_train.shape)
print("Testing input shape:", X_test.shape)
print("Testing target shape:", y_test.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping

vocab_size = len(word_to_id)

model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=64,
        mask_zero=True
    ),
    LSTM(128),
    Dense(vocab_size, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.SparseTopKCategoricalAccuracy(
            k=5,
            name="top5_accuracy"
        )
    ]
)

model.summary()
```

```python
import numpy as np
import re
from tensorflow.keras.preprocessing.sequence import pad_sequences

def predict_next_words(text, top_k=5):
    tokens = re.findall(r"\b[a-z]+(?:'[a-z]+)?\b", text.lower())

    token_ids = [
        word_to_id.get(word, word_to_id["<UNK>"])
        for word in tokens
    ]

    # Keep only the latest 19 tokens
    token_ids = token_ids[-(MAX_LEN - 1):]

    padded = pad_sequences(
        [token_ids],
        maxlen=MAX_LEN - 1,
        padding="pre",
        truncating="pre"
    )

    predictions = model.predict(padded, verbose=0)[0]

    # Exclude padding and unknown tokens
    predictions[0] = 0
    predictions[1] = 0

    top_indices = np.argsort(predictions)[-top_k:][::-1]

    print(f"\nInput phrase: {text}")
    print("Predicted next words:")

    for index in top_indices:
        print(
            f"{id_to_word[index]}: "
            f"{predictions[index] * 100:.2f}%"
        )


test_phrases = [
    "it was a",
    "she looked at",
    "there was no",
    "he said that",
    "the old man"
]

for phrase in test_phrases:
    predict_next_words(phrase)
```

```python
test_cases = [
    ("it was a", "little"),
    ("she looked at", "the"),
    ("there was no", "doubt"),
    ("he said that", "i"),
    ("the old man", "was")
]

correct_top1 = 0
correct_top5 = 0

for phrase, actual_word in test_cases:
    tokens = phrase.lower().split()

    token_ids = [
        word_to_id.get(word, word_to_id["<UNK>"])
        for word in tokens
    ]

    padded = pad_sequences(
        [token_ids[-(MAX_LEN - 1):]],
        maxlen=MAX_LEN - 1,
        padding="pre",
        truncating="pre"
    )

    predictions = model.predict(padded, verbose=0)[0]

    predictions[0] = 0
    predictions[1] = 0

    top5_ids = np.argsort(predictions)[-5:][::-1]
    predicted_words = [id_to_word[i] for i in top5_ids]

    top1_word = predicted_words[0]

    if top1_word == actual_word:
        correct_top1 += 1

    if actual_word in predicted_words:
        correct_top5 += 1

    print(f"\nInput: {phrase}")
    print(f"Actual next word: {actual_word}")
    print(f"Top prediction: {top1_word}")
    print(f"Top 5 predictions: {predicted_words}")
    print(f"Correct Top-1: {top1_word == actual_word}")
    print(f"Correct Top-5: {actual_word in predicted_words}")

print("\n========== TEST SUMMARY ==========")
print(f"Top-1 accuracy: {correct_top1 / len(test_cases) * 100:.2f}%")
print(f"Top-5 accuracy: {correct_top5 / len(test_cases) * 100:.2f}%")
```
